## 5. Data quality checks

This part of the notebook validates the integrity of the dataset by checking for missing values, duplicate rows, infinite values, and mixed data types. It also verifies whether every file contains the expected CICFlowMeter feature schema.

The goal is to identify issues that could distort training or reduce the reliability of the intrusion detection model.

```python
import pandas as pd
import numpy as np
from huggingface_hub import hf_hub_download

repo_id = "c01dsnap/CIC-IDS2018"

missing_results = []

for file_name in csv_files:
    print(f"\nChecking: {file_name}")

    path = hf_hub_download(
        repo_id=repo_id,
        filename=file_name,
        repo_type="dataset"
    )

    total_rows = 0
    total_missing = 0
    columns_with_missing = {}

    for chunk in pd.read_csv(path, chunksize=100_000):
        total_rows += len(chunk)

        missing_counts = chunk.isnull().sum()

        for column, count in missing_counts.items():
            if count > 0:
                columns_with_missing[column] = (
                    columns_with_missing.get(column, 0) + count
                )

        total_missing += missing_counts.sum()

    missing_results.append({
        "File": file_name,
        "Rows Checked": total_rows,
        "Total Missing Values": total_missing,
        "Columns With Missing": len(columns_with_missing)
    })

    print("Rows:", total_rows)
    print("Total missing values:", total_missing)
    print("Columns with missing values:", len(columns_with_missing))

    if columns_with_missing:
        print("Missing by column:")
        print(columns_with_missing)
    else:
        print("No missing values found.")

missing_df = pd.DataFrame(missing_results)

print("\n===== MISSING VALUE SUMMARY =====")
display(missing_df)
```

### Download dataset content

This cell downloads a CSV or metadata file from Hugging Face into the notebook environment so it can be inspected and analyzed.

```python
import pandas as pd
from huggingface_hub import hf_hub_download

duplicate_results = []

for file_name in csv_files:
    print(f"\nChecking: {file_name}")

    path = hf_hub_download(
        repo_id=repo_id,
        filename=file_name,
        repo_type="dataset"
    )

    total_rows = 0
    duplicate_rows = 0

    # Store hashes of rows already seen
    seen_hashes = set()

    for chunk in pd.read_csv(
        path,
        chunksize=100_000,
        low_memory=False
    ):
        total_rows += len(chunk)

        # Create a hash for each complete row
        row_hashes = pd.util.hash_pandas_object(
            chunk,
            index=False
        )

        # Count rows whose hash was already encountered
        for row_hash in row_hashes:
            if row_hash in seen_hashes:
                duplicate_rows += 1
            else:
                seen_hashes.add(row_hash)

    duplicate_percentage = (
        duplicate_rows / total_rows * 100
        if total_rows > 0 else 0
    )

    duplicate_results.append({
        "File": file_name,
        "Total Rows": total_rows,
        "Duplicate Rows": duplicate_rows,
        "Duplicate Percentage": duplicate_percentage
    })

    print("Total rows:", total_rows)
    print("Duplicate rows:", duplicate_rows)
    print(f"Duplicate percentage: {duplicate_percentage:.4f}%")

duplicate_df = pd.DataFrame(duplicate_results)

print("\n===== DUPLICATE ROW SUMMARY =====")
display(duplicate_df)
```

### Download dataset content

This cell downloads a CSV or metadata file from Hugging Face into the notebook environment so it can be inspected and analyzed.

```python
import pandas as pd
import numpy as np
from huggingface_hub import hf_hub_download

# ============================================================
# Combined Investigation: Steps 5.5 - 5.8
# ============================================================

# Columns that should normally be numeric
numeric_columns_expected = [
    "Dst Port",
    "Protocol",
    "Flow Duration",
    "Tot Fwd Pkts",
    "Tot Bwd Pkts",
    "TotLen Fwd Pkts",
    "TotLen Bwd Pkts",
    "Fwd Pkt Len Max",
    "Fwd Pkt Len Min",
    "Fwd Pkt Len Mean",
    "Fwd Pkt Len Std",
    "Bwd Pkt Len Max",
    "Bwd Pkt Len Min",
    "Bwd Pkt Len Mean",
    "Bwd Pkt Len Std",
    "Flow Byts/s",
    "Flow Pkts/s",
    "Flow IAT Mean",
    "Flow IAT Std",
    "Flow IAT Max",
    "Flow IAT Min",
    "Fwd IAT Tot",
    "Fwd IAT Mean",
    "Fwd IAT Std",
    "Fwd IAT Max",
    "Fwd IAT Min",
    "Bwd IAT Tot",
    "Bwd IAT Mean",
    "Bwd IAT Std",
    "Bwd IAT Max",
    "Bwd IAT Min",
    "Fwd PSH Flags",
    "Bwd PSH Flags",
    "Fwd URG Flags",
    "Bwd URG Flags",
    "Fwd Header Len",
    "Bwd Header Len",
    "Fwd Pkts/s",
    "Bwd Pkts/s",
    "Pkt Len Min",
    "Pkt Len Max",
    "Pkt Len Mean",
    "Pkt Len Std",
    "Pkt Len Var",
    "FIN Flag Cnt",
    "SYN Flag Cnt",
    "RST Flag Cnt",
    "PSH Flag Cnt",
    "ACK Flag Cnt",
    "URG Flag Cnt",
    "CWE Flag Count",
    "ECE Flag Cnt",
    "Down/Up Ratio",
    "Pkt Size Avg",
    "Fwd Seg Size Avg",
    "Bwd Seg Size Avg",
    "Fwd Byts/b Avg",
    "Fwd Pkts/b Avg",
    "Fwd Blk Rate Avg",
    "Bwd Byts/b Avg",
    "Bwd Pkts/b Avg",
    "Bwd Blk Rate Avg",
    "Subflow Fwd Pkts",
    "Subflow Fwd Byts",
    "Subflow Bwd Pkts",
    "Subflow Bwd Byts",
    "Init Fwd Win Byts",
    "Init Bwd Win Byts",
    "Fwd Act Data Pkts",
    "Fwd Seg Size Min",
    "Active Mean",
    "Active Std",
    "Active Max",
    "Active Min",
    "Idle Mean",
    "Idle Std",
    "Idle Max",
    "Idle Min"
]

# Features where negative values are potentially suspicious
non_negative_features = [
    "Dst Port",
    "Flow Duration",
    "Tot Fwd Pkts",
    "Tot Bwd Pkts",
    "TotLen Fwd Pkts",
    "TotLen Bwd Pkts",
    "Fwd Pkt Len Max",
    "Fwd Pkt Len Min",
    "Fwd Pkt Len Mean",
    "Bwd Pkt Len Max",
    "Bwd Pkt Len Min",
    "Bwd Pkt Len Mean",
    "Flow Byts/s",
    "Flow Pkts/s",
    "Flow IAT Mean",
    "Flow IAT Std",
    "Flow IAT Max",
    "Flow IAT Min",
    "Fwd Pkts/s",
    "Bwd Pkts/s",
    "Pkt Len Min",
    "Pkt Len Max",
    "Pkt Len Mean",
    "Pkt Len Std",
    "Pkt Len Var",
    "Active Mean",
    "Active Std",
    "Active Max",
    "Active Min",
    "Idle Mean",
    "Idle Std",
    "Idle Max",
    "Idle Min"
]

# ------------------------------------------------------------
# Result containers
# ------------------------------------------------------------

invalid_numeric_results = []
timestamp_results = []
label_results = []
range_results = []

# ============================================================
# Process every file
# ============================================================

for file_name in csv_files:

    print("\n" + "=" * 70)
    print("FILE:", file_name)
    print("=" * 70)

    path = hf_hub_download(
        repo_id=repo_id,
        filename=file_name,
        repo_type="dataset"
    )

    total_rows = 0

    # Step 5.5
    invalid_numeric_counts = {}

    # Step 5.6
    invalid_timestamp_count = 0
    missing_timestamp_count = 0
    min_timestamp = None
    max_timestamp = None

    # Step 5.7
    label_counts = {}

    # Step 5.8
    negative_counts = {}
    feature_min = {}
    feature_max = {}

    for chunk in pd.read_csv(
        path,
        chunksize=100_000,
        low_memory=False
    ):

        total_rows += len(chunk)

        # ====================================================
        # 5.5 Non-numeric / invalid values
        # ====================================================

        available_numeric_columns = [
            col for col in numeric_columns_expected
            if col in chunk.columns
        ]

        for col in available_numeric_columns:

            converted = pd.to_numeric(
                chunk[col],
                errors="coerce"
            )

            # Values that were originally non-null but
            # became NaN after numeric conversion
            invalid_mask = (
                chunk[col].notna()
                & converted.isna()
            )

            count = invalid_mask.sum()

            if count > 0:
                invalid_numeric_counts[col] = (
                    invalid_numeric_counts.get(col, 0) + count
                )

        # ====================================================
        # 5.6 Timestamp validity
        # ====================================================

        if "Timestamp" in chunk.columns:

            timestamps = pd.to_datetime(
                chunk["Timestamp"],
                format="%d/%m/%Y %H:%M:%S",
                errors="coerce"
            )

            missing_timestamp_count += chunk["Timestamp"].isna().sum()

            invalid_timestamp_count += (
                chunk["Timestamp"].notna()
                & timestamps.isna()
            ).sum()

            valid_timestamps = timestamps.dropna()

            if len(valid_timestamps) > 0:

                current_min = valid_timestamps.min()
                current_max = valid_timestamps.max()

                if min_timestamp is None or current_min < min_timestamp:
                    min_timestamp = current_min

                if max_timestamp is None or current_max > max_timestamp:
                    max_timestamp = current_max

        # ====================================================
        # 5.7 Label consistency
        # ====================================================

        if "Label" in chunk.columns:

            labels = chunk["Label"]

            counts = labels.value_counts(
                dropna=False
            )

            for label, count in counts.items():
                label_counts[label] = (
                    label_counts.get(label, 0) + count
                )

        # ====================================================
        # 5.8 Suspicious / extreme values
        # ====================================================

        available_non_negative = [
            col for col in non_negative_features
            if col in chunk.columns
        ]

        for col in available_non_negative:

            numeric_values = pd.to_numeric(
                chunk[col],
                errors="coerce"
            )

            negative_count = (
                numeric_values < 0
            ).sum()

            if negative_count > 0:
                negative_counts[col] = (
                    negative_counts.get(col, 0)
                    + negative_count
                )

            valid_values = numeric_values[
                np.isfinite(numeric_values)
            ].dropna()

            if len(valid_values) > 0:

                current_min = valid_values.min()
                current_max = valid_values.max()

                if (
                    col not in feature_min
                    or current_min < feature_min[col]
                ):
                    feature_min[col] = current_min

                if (
                    col not in feature_max
                    or current_max > feature_max[col]
                ):
                    feature_max[col] = current_max

    # ========================================================
    # Store results
    # ========================================================

    invalid_numeric_results.append({
        "File": file_name,
        "Rows": total_rows,
        "Columns With Non-Numeric Values": len(
            invalid_numeric_counts
        ),
        "Total Non-Numeric Values": sum(
            invalid_numeric_counts.values()
        )
    })

    timestamp_results.append({
        "File": file_name,
        "Rows": total_rows,
        "Missing Timestamps": missing_timestamp_count,
        "Invalid Timestamps": invalid_timestamp_count,
        "Minimum Timestamp": min_timestamp,
        "Maximum Timestamp": max_timestamp
    })

    label_results.append({
        "File": file_name,
        "Rows": total_rows,
        "Unique Labels": len(label_counts),
        "Label Counts": label_counts
    })

    range_results.append({
        "File": file_name,
        "Rows": total_rows,
        "Negative Value Columns": len(negative_counts),
        "Negative Value Counts": negative_counts,
        "Feature Minimums": feature_min,
        "Feature Maximums": feature_max
    })

    # ========================================================
    # Print file-level summary
    # ========================================================

    print("\n5.5 Non-numeric values:")
    print(
        "  Columns affected:",
        len(invalid_numeric_counts)
    )
    print(
        "  Total invalid numeric values:",
        sum(invalid_numeric_counts.values())
    )

    if invalid_numeric_counts:
        print("  Details:")
        print(invalid_numeric_counts)

    print("\n5.6 Timestamp:")
    print(
        "  Missing:",
        missing_timestamp_count
    )
    print(
        "  Invalid:",
        invalid_timestamp_count
    )
    print(
        "  Minimum:",
        min_timestamp
    )
    print(
        "  Maximum:",
        max_timestamp
    )

    print("\n5.7 Labels:")
    print("  Unique labels:", len(label_counts))
    print("  Labels:")
    for label, count in label_counts.items():
        print(f"    {repr(label)} : {count}")

    print("\n5.8 Negative values:")
    print(
        "  Columns with negative values:",
        len(negative_counts)
    )

    if negative_counts:
        print("  Details:")
        print(negative_counts)

# ============================================================
# Final summaries
# ============================================================

print("\n\n" + "=" * 80)
print("5.5 NON-NUMERIC VALUE SUMMARY")
print("=" * 80)

display(pd.DataFrame(invalid_numeric_results))

print("\n" + "=" * 80)
print("5.6 TIMESTAMP SUMMARY")
print("=" * 80)

display(pd.DataFrame(timestamp_results))

print("\n" + "=" * 80)
print("5.7 LABEL SUMMARY")
print("=" * 80)

for result in label_results:
    print("\n", result["File"])
    print(result["Label Counts"])

print("\n" + "=" * 80)
print("5.8 SUSPICIOUS / NEGATIVE VALUE SUMMARY")
print("=" * 80)

for result in range_results:
    print("\n", result["File"])

    print(
        "Negative-value columns:",
        result["Negative Value Columns"]
    )

    if result["Negative Value Counts"]:
        print(
            result["Negative Value Counts"]
        )
```

## 10. Temporal structure and time-window analysis

This section examines whether traffic is chronologically ordered and how attack and benign activity evolve over time. It parses the Timestamp field, groups flows into time windows, and identifies the dominant label within each window.

This helps determine whether the dataset contains recognizable temporal patterns and whether time-based modeling is suitable for attack forecasting or sequence analysis.

```python
import pandas as pd

path = file_path

window_size = "1min"

all_windows = []

for chunk in pd.read_csv(
    path,
    usecols=["Timestamp", "Label"],
    chunksize=200000
):

    # Remove header-like records
    chunk = chunk[chunk["Label"] != "Label"]

    # Parse timestamp
    chunk["Timestamp"] = pd.to_datetime(
        chunk["Timestamp"],
        format="%d/%m/%Y %H:%M:%S",
        errors="coerce"
    )

    chunk = chunk.dropna(subset=["Timestamp"])

    # Create 1-minute windows
    chunk["Time_Window"] = chunk["Timestamp"].dt.floor(window_size)

    # Count labels per window
    grouped = (
        chunk
        .groupby(["Time_Window", "Label"])
        .size()
        .reset_index(name="Count")
    )

    all_windows.append(grouped)

window_counts = pd.concat(
    all_windows,
    ignore_index=True
)

# Combine duplicate windows created by different chunks
window_counts = (
    window_counts
    .groupby(["Time_Window", "Label"])["Count"]
    .sum()
    .reset_index()
)

display(window_counts)
```

### Download dataset content

This cell downloads a CSV or metadata file from Hugging Face into the notebook environment so it can be inspected and analyzed.

```python
import pandas as pd
import numpy as np

dataset_files = [
    "Friday-02-03-2018_TrafficForML_CICFlowMeter.csv",
    "Friday-16-02-2018_TrafficForML_CICFlowMeter.csv",
    "Friday-23-02-2018_TrafficForML_CICFlowMeter.csv",
    "Thuesday-20-02-2018_TrafficForML_CICFlowMeter.csv",
    "Thursday-01-03-2018_TrafficForML_CICFlowMeter.csv",
    "Thursday-15-02-2018_TrafficForML_CICFlowMeter.csv",
    "Thursday-22-02-2018_TrafficForML_CICFlowMeter.csv",
    "Wednesday-14-02-2018_TrafficForML_CICFlowMeter.csv",
    "Wednesday-21-02-2018_TrafficForML_CICFlowMeter.csv",
    "Wednesday-28-02-2018_TrafficForML_CICFlowMeter.csv"
]

all_temporal_results = []

for filename in dataset_files:

    print("Processing:", filename)

    path = hf_hub_download(
        repo_id="c01dsnap/CIC-IDS2018",
        filename=filename,
        repo_type="dataset"
    )

    file_windows = []

    for chunk in pd.read_csv(
        path,
        usecols=["Timestamp", "Label"],
        chunksize=200000
    ):

        # Remove header-like records
        chunk = chunk[chunk["Label"] != "Label"]

        # Parse timestamp
        chunk["Timestamp"] = pd.to_datetime(
            chunk["Timestamp"],
            format="%d/%m/%Y %H:%M:%S",
            errors="coerce"
        )

        chunk = chunk.dropna(subset=["Timestamp"])

        # 5-minute temporal window
        chunk["Time_Window"] = (
            chunk["Timestamp"].dt.floor("5min")
        )

        # Count each label
        grouped = (
            chunk
            .groupby(["Time_Window", "Label"])
            .size()
            .reset_index(name="Count")
        )

        file_windows.append(grouped)

    # Combine chunk-level results
    file_windows = pd.concat(
        file_windows,
        ignore_index=True
    )

    # Combine repeated windows from different chunks
    file_windows = (
        file_windows
        .groupby(["Time_Window", "Label"])["Count"]
        .sum()
        .reset_index()
    )

    # Total flows per window
    total_counts = (
        file_windows
        .groupby("Time_Window")["Count"]
        .sum()
        .reset_index(name="Total_Flows")
    )

    # Attack flows
    attack_counts = (
        file_windows[
            file_windows["Label"] != "Benign"
        ]
        .groupby("Time_Window")["Count"]
        .sum()
        .reset_index(name="Attack_Flows")
    )

    # Merge
    temporal = total_counts.merge(
        attack_counts,
        on="Time_Window",
        how="left"
    )

    temporal["Attack_Flows"] = (
        temporal["Attack_Flows"]
        .fillna(0)
    )

    temporal["Benign_Flows"] = (
        temporal["Total_Flows"] -
        temporal["Attack_Flows"]
    )

    temporal["Attack_Percentage"] = (
        temporal["Attack_Flows"] /
        temporal["Total_Flows"]
    ) * 100

    # Dominant label
    dominant = (
        file_windows
        .sort_values(
            ["Time_Window", "Count"],
            ascending=[True, False]
        )
        .drop_duplicates("Time_Window")
        [["Time_Window", "Label"]]
    )

    dominant = dominant.rename(
        columns={"Label": "Dominant_Label"}
    )

    temporal = temporal.merge(
        dominant,
        on="Time_Window",
        how="left"
    )

    temporal["File"] = filename

    all_temporal_results.append(temporal)

temporal_all = pd.concat(
    all_temporal_results,
    ignore_index=True
)

print("Completed temporal investigation.")

display(temporal_all.head(30))
```

This code uploads a small verification file to the dataset repository to confirm that the connection and write permissions are working correctly.

```python
from huggingface_hub import HfApi

api = HfApi()

source_repo = "c01dsnap/CIC-IDS2018"
target_repo = "PrakulShetty/NetVision-AI-SIH26153"

files = [
    "Friday-02-03-2018_TrafficForML_CICFlowMeter.csv",
    "Friday-16-02-2018_TrafficForML_CICFlowMeter.csv",
    "Friday-23-02-2018_TrafficForML_CICFlowMeter.csv",
    "Thuesday-20-02-2018_TrafficForML_CICFlowMeter.csv",
    "Thursday-01-03-2018_TrafficForML_CICFlowMeter.csv",
    "Thursday-15-02-2018_TrafficForML_CICFlowMeter.csv",
    "Thursday-22-02-2018_TrafficForML_CICFlowMeter.csv",
    "Wednesday-14-02-2018_TrafficForML_CICFlowMeter.csv",
    "Wednesday-21-02-2018_TrafficForML_CICFlowMeter.csv",
    "Wednesday-28-02-2018_TrafficForML_CICFlowMeter.csv"
]

for file_name in files:
    print("Copying:", file_name)

    api.upload_file(
        path_or_fileobj=api.hf_hub_download(
            repo_id=source_repo,
            filename=file_name,
            repo_type="dataset"
        ),
        path_in_repo=file_name,
        repo_id=target_repo,
        repo_type="dataset"
    )

    print("Uploaded:", file_name)
```
